```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
def initialize_inventory(products):
    inventory = {}
    for product in products:
        quantity_of_each_product = int(input(f"Enter the quantity for {product}: "))
        inventory[product] = quantity_of_each_product
    return inventory
inventory = initialize_inventory(products)
```

```python
inventory
```

```python
def get_customer_orders():
    customer_orders = set()
    add_another = "yes"
    while add_another == "yes":
        name_of_product = input("Enter product name: ")
        customer_orders.add(name_of_product)
        add_another = input("Add another product? (yes/no): ").lower()
    return customer_orders
customer_orders = get_customer_orders()
```

```python
customer_orders
```

```python
def update_inventory(customer_orders,inventory):
    for product in customer_orders:
        inventory[product] = inventory[product] - 1
    return inventory
inventory = update_inventory(customer_orders, inventory)
```

```python
inventory
```

```python
def calculate_order_statistics(customer_orders,products):
    total_products_ordered = len(customer_orders)
    percentage_of_products_ordered = (len(customer_orders) / len(products)) * 100
    return total_products_ordered, percentage_of_products_ordered
order_statistics = calculate_order_statistics(customer_orders, products)
```

```python
order_statistics
```

```python
def print_order_statistics(order_statistics):
    print("Order Statistics:")
    print("Total Products Ordered:", order_statistics[0])
    print("Percentage of Products Ordered:", order_statistics[1])
```

```python
def print_updated_inventory(inventory):
    for key, value in inventory.items():
        print(key + ": " + str(value))
```

```python
inventory = initialize_inventory(products)
customer_orders = get_customer_orders()
order_statistics = calculate_order_statistics(customer_orders, products)
inventory = update_inventory(customer_orders, inventory)
print_order_statistics(order_statistics)
print_updated_inventory(inventory)
```
