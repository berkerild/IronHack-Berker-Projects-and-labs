```python
# Importing the libraries
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials

#Initialize SpotiPy with user credentials
sp = spotipy.Spotify(auth_manager=SpotifyClientCredentials(client_id=CLIENT_ID,
                                                           client_secret=CLIENT_SECRET))
```

```python
# Following the instructions

# - 1) List your favorite artists
artists = ["mor ve ötesi", "Jakuzi", "She Past Away", "Büyük Ev Ablukada", "Lalalar"]
```

```python
# - 2) Fetch top tracks
top_tracks_list = []

def get_top_tracks(artist_name):
    results = sp.search(q= artist_name, limit=10, type="track")
    for track in results["tracks"]["items"][:5]:
        top_tracks_list.append(track["name"])    
    return top_tracks_list
```

```python
# Calling the function for all the artists mentioned
for artist in artists:
    get_top_tracks(artist)

print(top_tracks_list)
```
