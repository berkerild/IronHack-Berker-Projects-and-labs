```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = {}

for product in products:
    quantity_of_each_product = int(input("Enter the quantity of product here: "))
    inventory[product] = quantity_of_each_product
print(inventory)
```
