```python
import numpy as np
import pandas as pd

insurance_df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv")
```

```python
insurance_df.head(10)
```

```python
insurance_clean_df = insurance_df.copy()
```

```python
# Challenge 1
# Exercise 1 

insurance_clean_df.columns = [column_name.lower() for column_name in insurance_clean_df.columns]
insurance_clean_df.columns
```

```python
insurance_clean_df.columns = [column_name.replace(" ", "_") for column_name in insurance_clean_df.columns]
insurance_clean_df.columns
```

```python
insurance_clean_df = insurance_clean_df.rename(columns={
    "st" : "state"
})
```

```python
insurance_clean_df.columns
```

```python
insurance_clean_df.head()
```

```python
# Exercise 2

for column in insurance_clean_df.columns:
    print(insurance_clean_df[column].value_counts())
    print()
```

```python
insurance_clean_df["gender"] = insurance_clean_df["gender"].replace({
    "Femal": "F",
    "female": "F",
    "Male": "M"
})
```

```python
insurance_clean_df["gender"].value_counts()
```

```python
insurance_clean_df["state"] = insurance_clean_df["state"].replace({
    "Cali": "California",
    "WA": "Washington",
    "AZ": "Arizona"
})
```

```python
insurance_clean_df["state"].value_counts()
```

```python
insurance_clean_df["education"] = insurance_clean_df["education"].replace({
    "Bachelors": "Bachelor"
})
```

```python
insurance_clean_df["education"].value_counts()
```

```python
insurance_clean_df["customer_lifetime_value"] = insurance_clean_df["customer_lifetime_value"].str.replace("%", "")
```

```python
insurance_clean_df["customer_lifetime_value"].head()
```

```python
insurance_clean_df["vehicle_class"] = insurance_clean_df["vehicle_class"].replace({
    "Sports Car": "Luxury",
    "Luxury SUV": "Luxury",
    "Luxury Car": "Luxury"
})
```

```python
insurance_clean_df["vehicle_class"].value_counts()
```

```python
# Exercise 3

insurance_clean_df.info()
```

```python
insurance_clean_df["customer_lifetime_value"] = insurance_clean_df["customer_lifetime_value"].astype("float")
```

```python
insurance_clean_df.info()
```

```python
insurance_clean_df["number_of_open_complaints"].unique()
```

```python
insurance_clean_df["number_of_open_complaints"] = insurance_clean_df["number_of_open_complaints"].str.split("/").str[1].astype(float)
```

```python
insurance_clean_df["number_of_open_complaints"].unique()
```

```python
# Exercise 4

insurance_clean_df.isnull().head()
```

```python
insurance_clean_df.isnull().sum()
```

```python
insurance_clean_df.shape
```

```python
insurance_clean_df.dropna().shape
```

```python
insurance_clean_df = insurance_clean_df.dropna(how= "all")
insurance_clean_df.shape
```

```python
insurance_clean_df.isnull().sum()
```

```python
most_common_gender = insurance_clean_df["gender"].mode()[0]
print("Most common gender:", most_common_gender)
```

```python
insurance_clean_df = insurance_clean_df.fillna({
    "gender": most_common_gender
})
```
