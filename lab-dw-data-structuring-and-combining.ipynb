```python
insurance_df_new = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv", parse_dates=["effective_to_date"])
insurance_df_new.info()
```

```python
insurance_df_new["sales_channel"].value_counts()
```

```python
# 1- Creating the pivot table showing the total revenue for each sales channel (branch, call center, web, and mail).

sales_channel_pivot = insurance_df_new.pivot_table(
    values="total_claim_amount", 
    index="sales_channel",
    aggfunc="sum"
)
```

```python
sales_channel_pivot = sales_channel_pivot.round(2)
sales_channel_pivot
```

```python
# Agent brings the most revenue at $1.81M, followed by Branch at $1.30M. Call Center and Web bring significantly less. This suggests the company relies heavily on human-driven sales channels, with Web being the least effective.
```

```python
# 2- Creating the pivot table that shows the average customer lifetime value per gender and education level. 

av_customer_lifetime_value = insurance_df_new.pivot_table(
    values="customer_lifetime_value", 
    index="gender",
    columns="education",
    aggfunc="mean"
)

av_customer_lifetime_value
```

```python
av_customer_lifetime_value = av_customer_lifetime_value.round(2)
av_customer_lifetime_value
```

```python
#Males with College education have the highest average CLV at $8,052.
#Females with High School or Below have the highest CLV among females at $8,675.
#Doctor level education has the lowest CLV for both genders.
#Overall CLV values are fairly similar across genders and education levels — no dramatic differences.

```
