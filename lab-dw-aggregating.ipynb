```python
import numpy as np
import pandas as pd

insurance_df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv")
print(insurance_df.info())
print()
insurance_df.head()
```

```python
insurance_df.columns = [column_name.lower() for column_name in insurance_df.columns]
insurance_df.columns
```

```python
insurance_df.columns = [column_name.replace(" ", "_") for column_name in insurance_df.columns]
insurance_df.columns
```

```python
# 1- Create a new data frame

insurance_df_new = insurance_df[
    (insurance_df["total_claim_amount"] < 1000) & (insurance_df["response"] == "Yes")
]

insurance_df_new.head()
```

```python
# 2. Using the original Dataframe, analyze:
   # the average `monthly_premium` and/or customer lifetime value by `policy_type` and `gender` for customers who responded "Yes", and
   # compare these insights to `total_claim_amount` patterns, and discuss which segments appear most profitable or low-risk for the company.

insurance_df_analysis = insurance_df.copy()
```

```python
insurance_df_analysis.head()
```

```python
condition = insurance_df_analysis["response"] == "Yes"
insurance_df_analysis = insurance_df_analysis[condition]
insurance_df_analysis.head()
```

```python
insurance_df_analysis["total_claim_amount"].value_counts()
```

```python
insurance_df_analysis = insurance_df_analysis.groupby(["gender", "policy_type"]).agg({
    "monthly_premium_auto" : "mean",
    "customer_lifetime_value" : "mean",
    "total_claim_amount" : "mean"
})
insurance_df_analysis
```

```python
# 3. Analyze the total number of customers who have policies in each state, and then filter the results to only include states where there are more than 500 customers.

insurance_df_state_policies = insurance_df.copy()
```

```python
insurance_df_state_policies["state"].value_counts()
```

```python
insurance_df_state_policies.groupby("state").size()
```

```python
state_counts = insurance_df_state_policies.groupby("state").size()
state_counts[state_counts > 500]
```

```python
# 4. Find the maximum, minimum, and median customer lifetime value by education level and gender. Write your conclusions.

insurance_df_math_calcs = insurance_df.groupby(["gender", "education"])["customer_lifetime_value"].agg(["max", "min", "median"])
insurance_df_math_calcs
```

```python
# Conclusions - 
# Male High School or Below has the highest max CLV at $83,325 — the most valuable individual customers come from this segment.
# Female Bachelor has the second highest max at $73,225.
# Male Doctor has the lowest max CLV at $32,677 — surprisingly, higher education doesn't mean higher value.
# Median values are fairly consistent across all segments, ranging from $5,332 to $6,286 — meaning the typical customer is similar regardless of gender or education.
# Minimum values are also very similar across all segments, around $1,900-$2,400.

```
