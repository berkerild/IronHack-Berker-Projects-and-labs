```python
insurance_clean_df = insurance_clean_df.rename(columns={
    "Customer": "customer",
    "ST": "st",
    "GENDER": "gender",
    "Education": "education",
    "Customer Lifetime Value": "clv",
    "Income": "income",
    "Monthly Premium Auto": "mpa",
    "Number of Open Complaints": "noc",
    "Policy Type": "pt",
    "Vehicle Class": "vc",
    "Total Claim Amount": "tca"
})

insurance_clean_df.columns
                                               
```

```python
insurance_clean_df["clv"] = insurance_clean_df["clv"].str.replace("%", "")
```

```python
# It looks like the data type for CLV and Number of Open Complaints are wrong. CLV should have been float and Number of Open Complaints have been integer.

insurance_clean_df["clv"] = insurance_clean_df["clv"].astype("float")
print(insurance_clean_df["clv"].dtypes)
```

```python
insurance_clean_df["noc"].unique()
```

```python
insurance_clean_df["noc"] = insurance_clean_df["noc"].str.split("/").str[1]
```

```python
insurance_clean_df["noc"] = insurance_clean_df["noc"].astype("float")
print(insurance_clean_df["noc"].dtypes)
```
