```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = {}

for product in products:
    quantity_of_each_product = int(input("Enter the quantity of product here: "))
    inventory[product] = quantity_of_each_product
print(inventory)
```

```python
customer_orders = set()

for product in range(3):
    name_of_three_products = str(input("Enter the name of product: "))
    customer_orders.add(name_of_three_products)
print(customer_orders)
```

```python
#Total Products Ordered
total_products_ordered = len(customer_orders)

#Percentage of Products Ordered
percentage_of_products_ordered = len(customer_orders) / len(products)

print(total_products_ordered)
print(percentage_of_products_ordered)
```

```python
order_status = (total_products_ordered, percentage_of_products_ordered)
print(order_status)
```

```python
print("Order Statistics:")
print("Total Products Ordered:", total_products_ordered)
print("Percentage of Products Ordered:", percentage_of_products_ordered)
```
